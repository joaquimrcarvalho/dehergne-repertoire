# Translate sources

This notebook prepares Kleio Sources for translation using AI code agents.

A given source can have the following content that requires translation:

1. Attributes (groups inheriting from group _attribute_):
   1. attribute types
   2. attribute values
2. Relations (groups inheriting from _relations_)
   1. relation types
   2. relation values
3. group names
4. elements in groups

For each of type of content this notebooks generates a list of strings in a file. It then facilitates the production of alternative files in other languages. From these set pf files new versions of the data are produced in diferent languages.

## Initialization

```python
from timelink.notebooks import TimelinkNotebook

tlnb = TimelinkNotebook()
tlnb.print_info(show_token=True)
```

## Step 1

Generate attribute type files

### Generate list of attribute types

```python
from sqlalchemy import select, func
from timelink.api.models.attribute import Attribute

# query unique values of field the_type of Attribute

stmt = select(Attribute.the_type).distinct().order_by(Attribute.the_type)
print(stmt)
with tlnb.db.session() as session:
    result = session.execute(stmt)

types = result.scalars().all()
for t in types:
    print(t)
types
```

#### Save file for future internationalization

1. Check if directory inferences/translations/attribute_types exists. If not create
2. Generate a file called attribute_type_base.csv with two columns in csv format, 
   each column containing the list of the original attribute types as fetched by
   the previous query. So both columns have identical values.


```python
from pathlib import Path
import csv

outdir = Path("../inferences/translations/attribute_types")
outdir.mkdir(parents=True, exist_ok=True)

outfile = outdir / "attribute_type_base.csv"
with outfile.open("w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["attribute_type", "attribute_type_copy"])  # header
    for t in types:
        writer.writerow([t, t])

print(f"Wrote {len(types)} rows to {outfile}")
```
